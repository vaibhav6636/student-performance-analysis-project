# 🔸 Outlier Detection & Handling (Numeric Columns)

### Objective:


```python
# Step 2: Outlier Detection & Handling
import numpy as np

numeric_cols = ['Percentage','study_hours','Sleep_Hours','Stress']  # important numeric columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(f"{col} - Outliers count: {len(outliers)}")
    
    # Handle by capping
    df[col] = np.where(df[col] < lower, lower, df[col])
    df[col] = np.where(df[col] > upper, upper, df[col])

print("\nOutliers handled for numeric columns.")
```

# 🔸 Correlation Analysis(Important Variables Only)

### Objective:
- To identify relationships between numerical variables.
- Positive values indicate direct relationships, while negative values indicate inverse relationships.

```python
plt.figure(figsize=(7,5))

# Select only important columns
cols = ['Percentage', 'study_hours', 'attendance', 
        'Sleep_Hours', 'Stress', 'Backlogs']

corr = df[cols].corr()

sns.heatmap(corr, annot=True, cmap='YlGn',fmt=".2f")

plt.title("Correlation Heatmap (Key Variables)")

plt.show()
```

### Interpretation

- The correlation analysis highlights that study hours and attendance have a positive association with academic performance, while stress shows a negative relationship. These findings reinforce the importance of consistent study habits and low stress levels for better academic outcomes.
